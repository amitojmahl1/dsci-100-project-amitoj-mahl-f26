#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# keep only the columns needed for the research question:
# seed and regular season win % (predictors) and tournament finish (what we predict)
march_madness_tidy <- march_madness |>
    select(seed, reg_wins_pct, tourney_finish) |>
    # remove seed 0, which was only used in 1983 for opening round games
    filter(seed != 0) |>
    # turn tournament finish into a categorical variable with levels in order
    mutate(tourney_finish = factor(tourney_finish,
                                   levels = c("opening_round_loss", "first_round_loss",
                                              "second_round_loss", "top_16_loss",
                                              "top_8_loss", "top_4_loss",
                                              "top_2_loss", "champ")))

# show the first 6 rows of the tidy data
head(march_madness_tidy)
```

```r
options(repr.plot.width = 10, repr.plot.height = 6)

# histogram of regular season win percentage
win_pct_hist <- march_madness_tidy |>
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(binwidth = 2, colour = "white") +
    labs(x = "Regular Season Win Percentage (%)",
         y = "Number of Teams",
         title = "Distribution of Regular Season Win Percentage") +
    theme(text = element_text(size = 16))
win_pct_hist
```

```r
# proportion of teams in each tournament finish
finish_props <- march_madness_tidy |>
    count(tourney_finish) |>
    mutate(proportion = n / sum(n))

finish_bar <- finish_props |>
    ggplot(aes(x = tourney_finish, y = proportion)) +
    geom_col() +
    labs(x = "Tournament Finish",
         y = "Proportion of Teams",
         title = "Proportion of Teams in Each Tournament Finish") +
    theme(text = element_text(size = 16),
          axis.text.x = element_text(angle = 45, hjust = 1))
finish_bar
```

One weakness is class imbalance: about 46% of teams are first round losses, while only about 2% are champions, so the model may mostly predict the common classes. KNN also only uses two predictors here, and seed has only 16 values, which can create ties between neighbours. It also does not explain why a team does well.

First, I will split the data into 75% training and 25% testing using initial_split, stratified by tourney_finish so each class shows up in both sets. I will create a recipe that centers and scales both predictors, since KNN uses distances and the variables are on different scales. On the training set, I will use 5-fold cross-validation to try K values from 1 to 50 and pick the K with the highest accuracy.
